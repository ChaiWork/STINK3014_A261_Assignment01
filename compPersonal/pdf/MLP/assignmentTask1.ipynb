## 4. Data Preprocessing & Splitting Pipeline

To guarantee rigorous empirical evaluation and zero data leakage:
1. **Feature Matrix ($X$) & Target Vector ($y$):** Partition attributes into strictly separated feature matrices and target vectors.
2. **Stratified Splitting:** Partition dataset into $70\%$ Training split and $30\%$ Held-out Test split.
3. **Inner Validation Split:** Further partition the $70\%$ training split into $80\%$ Inner Training ($56\%$ of total) and $20\%$ Validation ($14\%$ of total).
4. **Single StandardScaler Pipeline:** Compute $(\mu, \sigma)$ strictly on inner training data; transform validation and test sets using training statistics.


```
# ==============================================================================
# 4.1 Feature Matrix (X) and Target Vector (y) Construction
# ==============================================================================
X = df[input_features].copy()
y = df[target_col].copy().astype(int)
feature_cols = list(X.columns)

print("=== FEATURE & TARGET SPECIFICATION ===")
print(f"Total Records:           {len(df):,}")
print(f"Input Feature Matrix X:  {X.shape} ({len(feature_cols)} numerical attributes)")
print(f"Target Vector y:         {y.shape} (Binary class: 0=Benign, 1=Malicious)")
print(f"Class Proportions:       Benign: {(y==0).sum():,} ({(y==0).mean()*100:.2f}%) | Malicious: {(y==1).sum():,} ({(y==1).mean()*100:.2f}%)")
```

```
# ==============================================================================
# 4.2 Stratified Partitioning: 70% Train / 30% Test + 20% Inner Validation
# ==============================================================================
# Step 1: Split entire dataset into 70% Train and 30% Held-Out Test
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=SEED,
    stratify=y
)

# Step 2: Split 70% Train into 80% Inner Train and 20% Validation
X_tr_raw, X_val_raw, y_tr, y_val = train_test_split(
    X_train_raw, y_train,
    test_size=0.20,
    random_state=SEED,
    stratify=y_train
)

print("=== STRATIFIED DATASET PARTITIONING SUMMARY ===")
print(f"Total Dataset:           {len(df):,} records (100.0%)")
print(f"Training Split (70%):    {len(X_train_raw):,} records | Malicious Rate: {y_train.mean()*100:.2f}%")
print(f"  * Inner Train (80%):   {len(X_tr_raw):,} records | Malicious Rate: {y_tr.mean()*100:.2f}%")
print(f"  * Validation (20%):    {len(X_val_raw):,} records | Malicious Rate: {y_val.mean()*100:.2f}%")
print(f"Held-Out Test Set (30%): {len(X_test_raw):,} records | Malicious Rate: {y_test.mean()*100:.2f}%")
```

```
# ==============================================================================
# 4.3 Feature Standardization (Single StandardScaler fitted on Inner Train)
# ==============================================================================
scaler = StandardScaler()

# Fit scaler strictly on inner training split
X_tr_arr = scaler.fit_transform(X_tr_raw)
X_val_arr = scaler.transform(X_val_raw)
X_test_arr = scaler.transform(X_test_raw)

# Convert targets to numpy arrays
y_tr_arr = y_tr.to_numpy(dtype=np.float32)
y_val_arr = y_val.to_numpy(dtype=np.float32)
y_test_arr = y_test.to_numpy(dtype=np.float32)

# Verify parameters
scaler_param_df = pd.DataFrame({
    'Feature': feature_cols,
    'Train Mean (mu)': np.round(scaler.mean_, 4),
    'Train Std (sigma)': np.round(scaler.scale_, 4)
})

print("=== STANDARDSCALER PARAMETERS (Fitted on Training Set Only) ===")
display(scaler_param_df.head(10))
print(f"\nInput Tensors Ready: Train={X_tr_arr.shape}, Val={X_val_arr.shape}, Test={X_test_arr.shape}")
```

## 5. Task #1: Baseline MLP Model Construction & Backpropagation Training

### Architectural Design Specifications:
- **Input Dimension:** 10 standardized numerical features ($d_{in} = 13$)
- **Hidden Layer 1:** 32 neurons with Rectified Linear Unit (ReLU) activation ($g(z) = \max(0, z)$)
- **Hidden Layer 2:** 16 neurons with Rectified Linear Unit (ReLU) activation ($g(z) = \max(0, z)$)
- **Output Layer:** 1 neuron with Sigmoid activation ($\hat{y} = \sigma(z) = \frac{1}{1 + e^{-z}} \in [0, 1]$)
- **Loss Function:** Binary Cross-Entropy (BCE) Loss
- **Optimizer:** Adam (initial learning rate $\eta = 0.001$, $\beta_1 = 0.9$, $\beta_2 = 0.99$, $\epsilon = 10^{-7}$)
- **Batch Size:** 32 samples per mini-batch update
- **Training Duration:** 500 epochs


```
# ==============================================================================
# 5.1 Construct and Train Baseline MLP [13 -> 32 (ReLU) -> 16 (ReLU) -> 1 (Sigmoid)]
# ==============================================================================
import os
os.environ['KERAS_BACKEND'] = 'torch'
import keras
from keras import layers, models, optimizers

SEED = 42
keras.utils.set_random_seed(SEED)

# Build Sequential Architecture: [13 -> 32 -> 16 -> 1]
baseline_mlp = models.Sequential([
    layers.Input(shape=(X_tr_arr.shape[1],), name='input_features'),
    layers.Dense(32, activation='relu', name='hidden_layer_1_relu_32'),
    layers.Dense(16, activation='relu', name='hidden_layer_2_relu_16'),
    layers.Dense(1, activation='sigmoid', name='output_layer_sigmoid_1')
], name='Baseline_MLP_PDF_Phishing_Classifier')

# Compile with Binary Cross-Entropy loss and Adam optimizer (lr=0.001)
adam_opt = optimizers.Adam(learning_rate=0.001, beta_1=0.9, beta_2=0.99, epsilon=1e-7)
baseline_mlp.compile(
    optimizer=adam_opt,
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Print model architectural summary
baseline_mlp.summary()

# Train the model for 500 epochs with mini-batch size 32
print("\nStarting Baseline MLP Training (500 Epochs, Batch Size = 32, Adam Optimizer)...")
history = baseline_mlp.fit(
    X_tr_arr, y_tr_arr,
    validation_data=(X_val_arr, y_val_arr),
    epochs=500,
    batch_size=300,
    verbose=1
)
print("Baseline MLP training complete!")
```

```
# ==============================================================================
# 5.2 Baseline MLP Evaluation on Test Set & Learning Curve Visualization
# ==============================================================================
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report
)

# Predict probabilities and binary class on the held-out test set (30% split = 6,000 samples)
y_pred_probs = baseline_mlp.predict(X_test_arr, batch_size=64, verbose=0).flatten()
y_pred_class = (y_pred_probs >= 0.5).astype(int)

# Calculate quantitative test metrics
test_loss = baseline_mlp.evaluate(X_test_arr, y_test_arr, batch_size=64, verbose=0)[0]
test_acc = accuracy_score(y_test_arr, y_pred_class) * 100
test_prec = precision_score(y_test_arr, y_pred_class)
test_rec = recall_score(y_test_arr, y_pred_class)
test_f1 = f1_score(y_test_arr, y_pred_class)
test_auc = roc_auc_score(y_test_arr, y_pred_probs)

best_val_epoch = int(np.argmin(history.history['val_loss']) + 1)
min_val_loss = float(min(history.history['val_loss']))

print(f"=== BASELINE MLP HELD-OUT TEST EVALUATION (N={len(y_test_arr):,}) ===")
print(f"Test Loss:                 {test_loss:.4f}")
print(f"Test Accuracy:             {test_acc:.2f}%")
print(f"Test Precision:            {test_prec:.4f}")
print(f"Test Recall:               {test_rec:.4f}")
print(f"Test F1-Score:             {test_f1:.4f}")
print(f"Test ROC-AUC:              {test_auc:.4f}")
print(f"Best Validation Epoch:     Epoch {best_val_epoch} (Min Val Loss = {min_val_loss:.4f})")

print("\nClassification Report:")
print(classification_report(y_test_arr, y_pred_class, target_names=['Benign (0)', 'Malicious (1)']))

# Plot Training & Validation Learning Curves + Confusion Matrix
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Binary Cross-Entropy Loss Curves
epochs_range = range(1, len(history.history['loss']) + 1)
axes[0].plot(epochs_range, history.history['loss'], label='Training Loss', color='#2980b9', lw=1.8)
axes[0].plot(epochs_range, history.history['val_loss'], label='Validation Loss', color='#e74c3c', lw=1.8)
axes[0].axvline(best_val_epoch, color='gray', linestyle='--', alpha=0.7, label=f'Min Val Loss (Ep {best_val_epoch})')
axes[0].set_title('Baseline MLP: Binary Cross-Entropy Loss (500 Epochs)', fontweight='bold')
axes[0].set_xlabel('Epochs')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# 2. Accuracy (%) Convergence Curves
axes[1].plot(epochs_range, [a * 100 for a in history.history['accuracy']], label='Training Acc', color='#27ae60', lw=1.8)
axes[1].plot(epochs_range, [a * 100 for a in history.history['val_accuracy']], label='Validation Acc', color='#f39c12', lw=1.8)
axes[1].set_title('Baseline MLP: Accuracy (%) Convergence', fontweight='bold')
axes[1].set_xlabel('Epochs')
axes[1].set_ylabel('Accuracy (%)')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# 3. Test Set Confusion Matrix
cm = confusion_matrix(y_test_arr, y_pred_class)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, ax=axes[2],
            xticklabels=['Benign (0)', 'Malicious (1)'],
            yticklabels=['Benign (0)', 'Malicious (1)'],
            annot_kws={'size': 13, 'fontweight': 'bold'})
axes[2].set_title(f'Baseline MLP: Test Set Confusion Matrix (N={len(y_test_arr):,})', fontweight='bold')
axes[2].set_xlabel('Predicted Security Class')
axes[2].set_ylabel('Actual Ground Truth')

plt.tight_layout()
plt.show()
```

### Task #1 Analytical Questions & Answers

#### Question 1: What is the purpose of forward propagation and backpropagation?
- **Forward Propagation:** The computational process where an PDF input feature vector $\mathbf{x} \in \mathbb{R}^{10}$ flows forward across successive dense layers. At each layer $l$, affine combinations $z^{[l]} = W^{[l]} a^{[l-1]} + b^{[l]}$ are calculated and transformed via non-linear activations $a^{[l]} = g^{[l]}(z^{[l]})$. In the single-neuron output layer, the final logit is mapped to a calibrated probability estimate $\hat{y} = \sigma(z^{[L]}) \in [0, 1]$, enabling quantitative evaluation against the ground truth $y \in \{0, 1\}$ using Binary Cross-Entropy loss $\mathcal{L}(y, \hat{y})$.
- **Backpropagation:** The core training algorithm that applies the differential chain rule in reverse, transmitting error gradients from the output layer back to the input weights. It computes the exact partial derivatives $\frac{\partial \mathcal{L}}{\partial W^{[l]}}$ and $\frac{\partial \mathcal{L}}{\partial b^{[l]}}$, indicating how sensitively the loss changes with respect to each parameter to systematically reduce prediction error.

#### Question 2: How does the network update its weights to reduce prediction error?
Weights are updated in the direction opposite to the error gradient using gradient-based optimization:
$$W_{t+1} = W_t - \eta \nabla_{W} \mathcal{L}(W_t)$$
In the baseline Adam optimizer, the update incorporates exponentially decaying moving averages of past gradients (first moment $\hat{m}_t$, controlling momentum) and squared gradients (second moment $\hat{v}_t$, providing adaptive per-parameter learning rate scaling):
$$W_{t+1} = W_t - \frac{\eta}{\sqrt{\hat{v}_t} + \epsilon} \hat{m}_t$$
This ensures smooth, stable parameter updates across both dense count features and sparse ratio features.

#### Question 3: Why is binary cross-entropy suitable for this task?
PDF malicious webpage classification is a binary classification task ($y \in \{0, 1\}$). Binary Cross-Entropy (BCE) represents the negative log-likelihood under a Bernoulli distribution:
$$\mathcal{L}_{\text{BCE}}(y, \hat{y}) = - \left[ y \ln(\hat{y}) + (1 - y) \ln(1 - \hat{y}) \right]$$
1. **Asymmetric Penalty:** BCE severely penalizes high-confidence errors (as $\hat{y} \to 0$ when $y = 1$, $\mathcal{L} \to \infty$).
2. **Gradient Saturation Avoidance:** Combined with the sigmoid output $\hat{y} = \sigma(z)$, its derivative simplifies elegantly to $\frac{\partial \mathcal{L}}{\partial z} = \hat{y} - y$, preventing the gradient vanishing problems typical of Mean Squared Error (MSE) in classification.

#### Question 4: Why are ReLU and sigmoid suitable for the hidden and output layers, respectively?
- **Hidden Layers (ReLU):** $\text{ReLU}(z) = \max(0, z)$ provides a constant unit gradient of $1.0$ for all positive pre-activations ($z > 0$). This prevents vanishing gradients during multi-layer backpropagation, promotes sparse representations, and allows the network to partition the complex non-linear combinations of PDF attributes.
- **Output Layer (Sigmoid):** $\sigma(z) = \frac{1}{1 + e^{-z}}$ squashes real-valued logits $z \in (-\infty, \infty)$ into a bounded probability interval $[0, 1]$. In cybersecurity detection, this outputs the posterior probability $P(\text{Malicious} = 1 \mid \mathbf{x})$.

#### Question 5: What evidence in the learning curves suggests overfitting or underfitting?
- **Evidence of Overfitting:** In deep learning training across 500 epochs without regularization:
  1. **Loss Divergence:** Training loss continually declines towards zero, whereas validation loss reaches its global minimum at an early/intermediate epoch and subsequently begins drifting upwards.
  2. **Generalization Gap:** The divergence between training accuracy and validation accuracy demonstrates memorization of training-set noise, highlighting the necessity for regularisation (Task #4: L1/L2) and dropout (Task #5).
